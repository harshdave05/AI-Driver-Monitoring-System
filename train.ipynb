```python
# =============================================
# 🧠 YOLOv8 Multi-Model Training Pipeline
# Models: Face Mask, Hand, Smoke Detection
# =============================================

!pip install ultralytics --quiet

from ultralytics import YOLO
import os

# ======================
# ⚙️ Check environment
# ======================
import torch
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f"Using device: {device}")


FACE_MASK_YAML = "From_kaggle/face_mask/data.yaml"
HAND_YAML = "From_roboflow_datasets/hand/data.yaml"
SMOKE_YAML = "datasets/smoke/data.yaml"

# ======================
# Train Function
# ======================
def train_yolo_model(dataset_yaml, model_name, epochs=100, img_size=640):
    """
    Train a YOLOv8 model on a given dataset.
    """
    print(f"\n Training model: {model_name}")
    model = YOLO("yolov8n.pt")  # start from YOLOv8 nano (you can use yolov8s.pt for higher accuracy)
    
    model.train(
        data=dataset_yaml,
        epochs=epochs,
        imgsz=img_size,
        batch=16,
        device=device,
        name=model_name,
        project="runs/train",
        workers=4
    )

    # Export best model
    best_path = f"runs/train/{model_name}/weights/best.pt"
    if os.path.exists(best_path):
        print(f"Model saved at: {best_path}")
    else:
        print(" Training completed but best.pt not found.")

# ======================
#  Train All Models
# ======================

# 1️⃣ Face Mask Detection
train_yolo_model(FACE_MASK_YAML, model_name="face_mask_detector", epochs=100)

# 2️⃣ Hand Detection
train_yolo_model(HAND_YAML, model_name="hand_detector", epochs=100)

# 3️⃣ Smoke Detection
train_yolo_model(SMOKE_YAML, model_name="smoke_detector", epochs=100)

print("\n All training processes completed successfully!")
```
